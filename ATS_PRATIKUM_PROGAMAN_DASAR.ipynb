```python
# SOAL NOMOR 2
# a. Simpan nilai dalam variabel dan hitung energi harian
daya_kw = 450
jam = 8
energi = daya_kw * jam

print(f"Daya PLTD        : {daya_kw} kW")
print(f"Lama operasi     : {jam} jam")
print(f"Energi harian    : {energi} kWh")

# b. Hitung capacity factor
energi_aktual = 7200
kapasitas_kw = 400
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Energi aktual    : {energi_aktual} kWh")
print(f"Kapasitas        : {kapasitas_kw} kW")
print(f"Capacity factor  : {capacity_factor:.2f} %")
```

```python
# SOAL NOMOR 10
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for n, teks in enumerate(data, start=1):
    # a. Ubah ke float dengan try-except
    try:
        suhu = float(teks)
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break

    # Data berhasil dikonversi, hitung sebagai data valid
    jumlah_valid += 1

    # b. Cek suhu tinggi
    if suhu > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")

# c. Tampilkan jumlah data valid setelah perulangan selesai
print(f"Jumlah data valid: {jumlah_valid}")
```

```python
# ==========================================================
# PEMANTAUAN BEBAN HARIAN PLTD
#
# Masalah:
#   Operator PLTD perlu memantau beban listrik tiap jam, menandai
#   kondisi beban (rendah/normal/siaga/overload), dan menghitung
#   total energi, beban puncak, load factor, serta estimasi biaya.
#   Sensor bisa rusak, sehingga program harus tetap berjalan.
#
# Cara kerja:
#   1. Data beban (teks dari sensor) disimpan dalam list.
#   2. Perulangan for memeriksa data satu per satu.
#   3. try-except menangani data rusak (dilewati dengan continue).
#   4. if-elif-else menentukan status beban dari persentase kapasitas.
#   5. Setelah loop, program menampilkan ringkasan hasil.
# ==========================================================

# --- Data dan parameter ---
kapasitas_kw = 500          # kapasitas terpasang (kW)
tarif_per_kwh = 1500        # estimasi biaya bahan bakar (Rp/kWh)
data_sensor = ["220", "185", "170", "190", "260", "340",
               "410", "455", "N/A", "520", "480", "390"]  # 1 pembacaan = 1 jam

beban_valid = []            # list beban (kW) yang berhasil dibaca
status_list = []            # list status tiap beban valid
beban_puncak = 0
jam_puncak = 0

print("LAPORAN BEBAN HARIAN PLTD")
print("-" * 46)

# --- Perulangan pemeriksaan data ---
for jam, teks in enumerate(data_sensor, start=1):
    # Konversi teks ke float; data rusak/negatif dilewati
    try:
        beban = float(teks)
        if beban < 0:
            raise ValueError("beban negatif")
    except ValueError:
        print(f"Jam {jam:>2} | data sensor tidak valid ({teks!r}) - dilewati")
        continue

    # Hitung persentase beban terhadap kapasitas
    persen = beban / kapasitas_kw * 100

    # Tentukan status beban
    if persen >= 100:
        status = "OVERLOAD"
    elif persen >= 80:
        status = "SIAGA"
    elif persen >= 40:
        status = "NORMAL"
    else:
        status = "RENDAH"

    # Simpan data dan perbarui beban puncak
    beban_valid.append(beban)
    status_list.append(status)
    if beban > beban_puncak:
        beban_puncak = beban
        jam_puncak = jam

    print(f"Jam {jam:>2} | {beban:>6.1f} kW | {persen:>5.1f}% | {status}")

# --- Ringkasan hasil ---
print("-" * 46)
if len(beban_valid) == 0:
    print("Tidak ada data valid untuk dianalisis.")
else:
    total_energi = sum(beban_valid)                    # kWh (tiap data = 1 jam)
    rata_rata = total_energi / len(beban_valid)        # kW
    load_factor = rata_rata / beban_puncak * 100       # %
    biaya = total_energi * tarif_per_kwh               # Rp

    print(f"Jumlah data valid : {len(beban_valid)} dari {len(data_sensor)}")
    print(f"Total energi      : {total_energi:.1f} kWh")
    print(f"Rata-rata beban   : {rata_rata:.2f} kW")
    print(f"Beban puncak      : {beban_puncak:.1f} kW (jam ke-{jam_puncak})")
    print(f"Load factor       : {load_factor:.2f} %")
    print(f"Estimasi biaya    : Rp {biaya:,.0f}".replace(",", "."))

    print("\nDistribusi status:")
    for s in ["RENDAH", "NORMAL", "SIAGA", "OVERLOAD"]:
        print(f"  {s:<8}: {status_list.count(s)} jam")

    if status_list.count("OVERLOAD") > 0:
        print("\nREKOMENDASI: Terjadi overload, pertimbangkan menambah unit genset.")
    elif status_list.count("SIAGA") > 0:
        print("\nREKOMENDASI: Beban mendekati batas, siapkan unit cadangan.")
    else:
        print("\nREKOMENDASI: Operasi aman.")
```
